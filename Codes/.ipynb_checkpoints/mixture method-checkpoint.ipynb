```python
import numpy as np
from sklearn.linear_model import Ridge
from scipy.special import expit
import matplotlib.pyplot as plt

# 固定随机种子，保证每次运行结果完全一致
# Fix random seed to ensure identical results every run
np.random.seed(42)
```

```python
# 步骤 1：生成最简单的模拟数据（先搞懂数据长什么样）
# Step 1: Generate the simplest simulated data (first understand what the data looks like)
# 生成2个站点的逻辑回归数据，每个站点有100个样本、10个特征，其中前2个是真实相关特征，后8个是噪声特征。
# Generate logistic regression data for 2 sites, each with 100 samples and 10 features, where the first 2 are true relevant features and the last 8 are noise features.
```

```python
def generate_one_site(n, p, beta_true):
    # 生成特征：简单的独立正态分布（不用复杂的Toeplitz相关）
    # Generate features: simple independent normal distribution (no complex Toeplitz correlation)
    X = np.random.randn(n, p)
    
    # 生成响应变量y（0/1） 
    # Generate response variable y (0/1)
    y = np.random.binomial(1, expit(X @ beta_true))
    
    return X, y

# 生成2个站点数据 
# Generate data for 2 sites
n = 100  # 每个站点100个样本 | 100 samples per site
p = 10   # 总特征数 | Total number of features
p1 = 2   # 真实相关特征数（前2个） | Number of true relevant features (first 2)
signal_strength = 2.0

# 关键修改：只生成一次公共的beta_true，两个站点共用 
# Key modification: generate public beta_true once, shared by both sites
beta_true = np.zeros(p)
beta_true[:p1] = signal_strength * np.random.choice([-1, 1], p1)
print(f"公共真实系数beta_true: \nCommon true coefficients beta_true: {beta_true.round(2)}")

site1_X, site1_y = generate_one_site(n, p, beta_true)
site2_X, site2_y = generate_one_site(n, p, beta_true)

print("=== 数据结构 ===  \n=== Data Structure ===")
print(f"站点1特征矩阵形状: \nSite 1 feature matrix shape: {site1_X.shape}")
print(f"站点1响应变量形状: \nSite 1 response variable shape: {site1_y.shape}")
print(f"真实相关特征索引: \nTrue relevant feature indices: {list(range(p1))}")
```

```python
# 步骤 2：全局数据诊断（无隐私泄露） 
# Step 2: Global Data Diagnosis (No Privacy Leakage)
# 这一步做什么 
# What this step does
# 根据每个站点的样本量，判断它应该走路线 A（低样本：样本切分）还是路线 B（高样本：全量数据）。 
# Based on the sample size of each site, determine whether it should take Route A (low sample: sample splitting) or Route B (high sample: full data).
```

```python
# 理论阈值：当样本量 >= C*p1*log(p) 时，全量数据的选择偏差可被DP噪声抵消 
# Theoretical threshold: when sample size >= C*p1*log(p), selection bias of full data can be offset by DP noise
C = 2.0
n_thresh = C * p1 * np.log(p)
print(f"样本量阈值n_thresh = {n_thresh:.1f} | Sample size threshold n_thresh = {n_thresh:.1f}")

# 两个站点的样本量都是100，都大于阈值，所以都走路线B 
# Both sites have sample size 100, which is greater than the threshold, so both take Route B
site1_n = site1_X.shape[0]
site2_n = site2_X.shape[0]

print(f"\n站点1样本量：{site1_n} → 高样本，走路线B | Site 1 sample size: {site1_n} → High sample, take Route B")
print(f"站点2样本量：{site2_n} → 高样本，走路线B | Site 2 sample size: {site2_n} → High sample, take Route B")
```

```python
# 为什么这么做 
# Why we do this
# 当样本量足够大时，全量数据的选择偏差很小，DP 噪声可以完全抵消 
# When the sample size is large enough, the selection bias of full data is small and can be completely offset by DP noise
# 当样本量很小时，必须用样本切分来保证 FDR 控制的严格性 
# When the sample size is small, sample splitting must be used to ensure the rigor of FDR control
```

```python
# 步骤 3：本地差分隐私粗筛（路线 B 核心）
# Step 3: Local Differential Privacy Screening (Core of Route B)
# 这一步做什么 
# What this step does
# 每个站点用全量数据计算每个特征与响应的边际相关性，然后加入 GDP 噪声，本地选出前 d 个特征作为本地支持集。
# Each site uses full data to calculate the marginal correlation between each feature and the response, then adds GDP noise, and selects the top d features locally as the local support set.
```

```python
def local_dp_sis(X, y, mu1, d):
    n = X.shape[0]
    p = X.shape[1]
    
    # 步骤1：计算边际相关性得分 r_j = (X_j^T y)/n 
    # Step 1: Calculate marginal correlation score r_j = (X_j^T y)/n
    # 这个得分反映了特征j与响应y的关联强度 | This score reflects the association strength between feature j and response y
    r = X.T @ y / n
    print(f"原始边际相关性得分:  \nOriginal marginal correlation scores: \n{r.round(3)}")
    
    # 步骤2：计算全局敏感度（边际相关性的最大可能变化量）
    # Step 2: Calculate global sensitivity (maximum possible change in marginal correlation)
    # 当改变一个样本时，r_j的变化量最大为 2/n | When changing one sample, the maximum change in r_j is 2/n
    delta_r = 2.0 / n
    print(f"边际相关性敏感度:  \nMarginal correlation sensitivity: \n{delta_r:.3f}")
    
    # 步骤3：加入μ1-GDP噪声 
    # Step 3: Add μ1-GDP noise
    # GDP噪声标准差公式：sigma = delta * 1.25 / mu | GDP noise standard deviation formula: sigma = delta * 1.25 / mu
    sigma_r = delta_r * 1.25 / mu1
    r_noisy = r + np.random.normal(0, sigma_r, p)
    print(f"加噪后的边际相关性得分: \nNoisy marginal correlation scores: \n{r_noisy.round(3)}")
    
    # 步骤4：本地选前d个特征（d=2*p1，保证不丢真实特征） 
    # Step 4: Select top d features locally (d=2*p1 to ensure no true features are missed)
    top_indices = np.argsort(np.abs(r_noisy))[-d:]
    local_support = set(top_indices)
    
    return local_support, r_noisy

# 粗筛阶段隐私预算 
# Privacy budget for screening stage
mu1 = 0.5
# 每个站点选前4个特征（2倍真实特征数）
# Select top 4 features per site (2x the number of true features)
d = 2 * p1

print("=== 站点1本地粗筛 ===  \n=== Site 1 Local Screening ===")
site1_support, site1_r_noisy = local_dp_sis(site1_X, site1_y, mu1, d)
print(f"站点1本地支持集: \nSite 1 local support set:\n {site1_support}")

print("\n=== 站点2本地粗筛 ===  \n=== Site 2 Local Screening ===")
site2_support, site2_r_noisy = local_dp_sis(site2_X, site2_y, mu1, d)
print(f"站点2本地支持集:  \nSite 2 local support set:\n {site2_support}")
```

```python
# 这里是第一个隐私泄露点：如果直接上传原始边际相关性得分，攻击者可以反推原始数据
# This is the first privacy leakage point: if the original marginal correlation scores are uploaded directly, attackers can reverse-engineer the original data
# 我们加入了 GDP 噪声，使得攻击者无法区分两个相邻数据集的输出，从而保护了隐私 
# We add GDP noise so that attackers cannot distinguish the outputs of two adjacent datasets, thus protecting privacy
# 噪声的大小由隐私预算 μ1 决定：μ1 越小，噪声越大，隐私保护越强，但效用越差 
# The magnitude of noise is determined by the privacy budget μ1: the smaller μ1 is, the larger the noise, the stronger the privacy protection, but the worse the utility
```

```python
# 步骤 4：安全交换与多数表决（解决假阳性爆炸）
# Step 4: Secure Exchange and Majority Voting (Solve False Positive Explosion)
# 这一步做什么
# What this step does
# 两个站点互相交换本地支持集，然后执行多数表决：只有在两个站点的支持集中都出现的特征，才能进入最终的鲁棒支持集。
# The two sites exchange local support sets with each other, then perform majority voting: only features that appear in both sites' support sets can enter the final robust support set.
```

```python
# 站点1和站点2互相交换本地支持集 
# Site 1 and Site 2 exchange local support sets with each other
# 这里是第二个隐私泄露点：如果直接交换真实支持集，会泄露本地数据的特征分布 
# This is the second privacy leakage point: if the true support sets are exchanged directly, the feature distribution of local data will be leaked
# 实际应用中需要对支持集加轻量级DP保护（如随机响应），这里简化处理 
# In practical applications, lightweight DP protection (e.g., random response) needs to be added to the support sets, which is simplified here
print("=== 多数表决 ===\n=== Majority Voting ===")
print(f"站点1支持集: \nSite 1 support set: {site1_support}")
print(f"站点2支持集: \nSite 2 support set: {site2_support}")

# 简单多数表决：得票≥1票（两个站点都出现）
# Simple majority voting: votes ≥ 1 (appears in both sites)
S_robust = site1_support & site2_support
print(f"鲁棒支持集（交集）: \nRobust support set (intersection): {S_robust}")
```

```python
# 为什么这么做 
# Why we do this
# 原 Fed-FDR 的跨站点取并集会导致假阳性 "滚雪球"：一个站点的假阳性会被所有站点继承 
# The union across sites in original Fed-FDR causes a false positive "snowball effect": a false positive from one site is inherited by all sites
# 多数表决可以过滤掉大部分只在单个站点出现的假阳性特征 
# Majority voting filters out most false positive features that appear in only a single site
# 这里用交集（得票率 100%）是最严格的情况，实际中可以用 50% 的简单多数 
# Using intersection (100% vote rate) is the strictest case here; in practice, 50% simple majority can be used
```

```python
# 步骤 5：私有化低维估计（核心隐私保护）
# Step 5: Privatized Low-dimensional Estimation (Core Privacy Protection)
# 这一步做什么 
# What this step does
# 每个站点用全量数据在鲁棒支持集上运行低维 Ridge 回归，然后加入 GDP 噪声，将私有化后的系数上传到中心。
# Each site runs low-dimensional Ridge regression on the robust support set using full data, then adds GDP noise and uploads the privatized coefficients to the center.
```

```python
def privatized_low_dim_estimation(X, y, features, mu2):
    n = X.shape[0]
    # 只保留鲁棒支持集中的特征 
    # Only retain features in the robust support set
    X_sub = X[:, list(features)]
    
    # 步骤1：低维Ridge回归 
    # Step 1: Low-dimensional Ridge regression
    ridge = Ridge(alpha=1e-5)
    ridge.fit(X_sub, y)
    beta_hat = ridge.coef_
    print(f"原始Ridge系数: \nOriginal Ridge coefficients: {beta_hat.round(3)}")
    
    # 步骤2：计算系数的全局敏感度 
    # Step 2: Calculate global sensitivity of coefficients
    # 低维Ridge系数的敏感度 = 2 / (n * 最小特征值)
    # Sensitivity of low-dimensional Ridge coefficients = 2 / (n * minimum eigenvalue)
    cov_sub = X_sub.T @ X_sub / n
    min_eig = np.linalg.eigvalsh(cov_sub)[0]
    delta_beta = 2.0 / (n * min_eig)
    print(f"系数敏感度: \nCoefficient sensitivity: {delta_beta:.3f}")
    
    # 步骤3：加入μ2-GDP噪声 
    # Step 3: Add μ2-GDP noise
    sigma_beta = delta_beta * 1.25 / mu2
    beta_noisy = beta_hat + np.random.normal(0, sigma_beta, len(features))
    print(f"加噪后的Ridge系数: \nNoisy Ridge coefficients: {beta_noisy.round(3)}")
    
    # 步骤4：构造完整系数向量（未选中特征为0）
    # Step 4: Construct full coefficient vector (unselected features are 0)
    beta_full = np.zeros(p)
    beta_full[list(features)] = beta_noisy
    
    return beta_full

# 估计阶段隐私预算 
# Privacy budget for estimation stage
mu2 = 1.0

print("\n=== 站点1私有化估计 ===\n=== Site 1 Privatized Estimation ===")
site1_beta_noisy = privatized_low_dim_estimation(site1_X, site1_y, S_robust, mu2)

print("\n=== 站点2私有化估计 ===\n=== Site 2 Privatized Estimation ===")
site2_beta_noisy = privatized_low_dim_estimation(site2_X, site2_y, S_robust, mu2)

# 中心收到两个站点的私有化系数 | Center receives privatized coefficients from both sites
beta_noisy = np.array([site1_beta_noisy, site2_beta_noisy])
print(f"\n中心收到的系数矩阵形状: \nShape of coefficient matrix received by center: {beta_noisy.shape}")
```

```python
# 关键隐私保护说明
# Key Privacy Protection Explanation
# 这里是最严重的隐私泄露点：原始 Ridge 系数直接反映了本地数据中特征与响应的关联强度
# This is the most serious privacy leakage point: the original Ridge coefficients directly reflect the correlation strength between features and responses in local data
# 我们加入了 GDP 噪声，使得攻击者无法从加噪后的系数反推原始数据 
# We add GDP noise so that attackers cannot reverse-engineer the original data from the noisy coefficients
# 由于我们只在低维特征上做估计，敏感度远低于高维情况，所以噪声可以加得很小 
# Since we only perform estimation on low-dimensional features, the sensitivity is much lower than that in high-dimensional cases, so the noise can be very small
```

```python
# 步骤 6：中心计算广义镜像统计量 
# Step 6: Center Computes Generalized Mirror Statistics
# 这一步做什么
# What this step does
# 中心用两个站点上传的已经加过噪的系数，计算广义镜像统计量。
# The center uses the noisy coefficients uploaded by the two sites to compute the generalized mirror statistics.
```

```python
def compute_mirror_stats(beta_noisy):
    """计算广义镜像统计量  \nCompute generalized mirror statistics"""
    K = beta_noisy.shape[0]
    p = beta_noisy.shape[1]
    M = np.zeros(p)
    
    # 对每一对站点计算镜像统计量 
    # Calculate mirror statistics for each pair of sites
    for s in range(K):
        for t in range(s+1, K):
            beta_s = beta_noisy[s]
            beta_t = beta_noisy[t]
            # 广义镜像统计量公式：M_j = sign(beta_sj * beta_tj) * (|beta_sj| + |beta_tj|)
            # Generalized mirror statistics formula: M_j = sign(beta_sj * beta_tj) * (|beta_sj| + |beta_tj|)
            sign = np.sign(beta_s * beta_t)
            abs_sum = np.abs(beta_s) + np.abs(beta_t)
            M += sign * abs_sum
    
    # 平均所有站点对的结果 
    # Average the results across all site pairs
    M = M / (K*(K-1)/2)
    return M

# 计算镜像统计量 
# Compute mirror statistics
M = compute_mirror_stats(beta_noisy)
print(f"广义镜像统计量:  \nGeneralized mirror statistics: {M.round(3)}")

# 解决matplotlib中文显示问题 
# Solve matplotlib Chinese display problem
plt.rcParams['font.sans-serif'] = ['SimHei']  # Windows用黑体 
# Use SimHei font for Windows
plt.rcParams['axes.unicode_minus'] = False    # 解决负号显示为方框的问题 
# Fix the issue of minus signs displaying as boxes

# 可视化镜像统计量分布 
# Visualize the distribution of mirror statistics
plt.figure(figsize=(10, 4))
plt.bar(range(p), M, color=['red' if i < p1 else 'blue' for i in range(p)])
plt.axhline(0, color='black', linestyle='--')
plt.xlabel('特征索引 | Feature Index')
plt.ylabel('镜像统计量M_j | Mirror Statistics M_j')
plt.title('镜像统计量分布（红色=真实特征，蓝色=噪声特征）| Distribution of Mirror Statistics (Red=True Features, Blue=Noise Features)')
plt.show()
```

```python
# 步骤 7：FDR 控制与结果评估 
# Step 7: FDR Control and Result Evaluation
# 这一步做什么 
# What this step does
# 中心根据镜像统计量的分布，选择合适的阈值，控制 FDR 在目标水平。
# The center selects an appropriate threshold based on the distribution of mirror statistics to control FDR at the target level.
```

```python
# ==============================================
# ✅ 修正后的A-B混合融合FDR控制（真正的融合方法）
# ==============================================
def hybrid_fdr_control(M, alpha):
    """
    我们提出的A-B混合FDR控制法 | Our proposed A-B hybrid FDR control method
    同时运行路线A和路线B，取并集作为最终结果 | Run both Route A and Route B, take union as final result
    """
    print("\n=== 路线A（正侧FDR）阈值搜索 === \n === Route A (Positive-side FDR) Threshold Search ===")
    # --------------------------
    # 第一步：独立运行路线A（正侧FDR）
    # --------------------------
    thresholds_A = np.sort(M)[::-1]
    best_t_A = 0
    for t in thresholds_A:
        if t <= 0:
            continue
        num_neg = np.sum(M < -t)
        num_pos = np.sum(M > t)
        fdp_est_A = num_neg / max(num_pos, 1)
        print(f"  阈值t={t:.3f}，估计FDP={fdp_est_A:.3f}\nThreshold t={t:.3f}, estimated FDP={fdp_est_A:.3f}")
        if fdp_est_A <= alpha:
            best_t_A = t
            break
    S_A = set(np.where(M > best_t_A)[0])
    
    print("\n=== 路线B（绝对值FDR）阈值搜索 ===\n === Route B (Absolute-value FDR) Threshold Search ===")
    # --------------------------
    # 第二步：独立运行路线B（绝对值FDR）
    # --------------------------
    thresholds_B = np.sort(np.abs(M))[::-1]
    best_t_B = 0
    for t in thresholds_B:
        if t == 0:
            continue
        num_neg = np.sum(M < -t)
        num_pos = np.sum(M > t)
        fdp_est_B = 2 * min(num_neg, num_pos) / max(num_pos + num_neg, 1)
        print(f"  阈值t={t:.3f}，估计FDP={fdp_est_B:.3f} \nThreshold t={t:.3f}, estimated FDP={fdp_est_B:.3f}")
        if fdp_est_B <= alpha:
            best_t_B = t
            break
    S_B = set(np.where((M >= best_t_B) | (M <= -best_t_B))[0])
    
    # --------------------------
    # 第三步：融合结果（取并集）
    # --------------------------
    S_final = S_A.union(S_B)
    return S_final, (best_t_A, best_t_B)

# 目标FDR水平 | Target FDR level
alpha = 0.1

print("=== 混合FDR控制 ===\n === Hybrid FDR Control ===")
hat_S, (t_A, t_B) = hybrid_fdr_control(M, alpha)
print(f"\n路线A阈值t_A={t_A:.3f}\n Route A threshold t_A={t_A:.3f}")
print(f"路线B阈值t_B={t_B:.3f}\n Route B threshold t_B={t_B:.3f}")
print(f"最终选中特征集：{hat_S}\n Final selected feature set: {hat_S}")

# 结果评估 | Result Evaluation
S1 = set(range(p1))
TP = len(hat_S & S1)
FP = len(hat_S - S1)
FDR = FP / max(len(hat_S), 1)
Power = TP / len(S1)

print(f"\n=== 最终结果 ===\n === Final Results ===")
print(f"真实相关特征：{S1}\n True relevant features: {S1}")
print(f"选中特征：{hat_S}\n Selected features: {hat_S}")
print(f"真实FDR：{FDR:.3f}（目标：{alpha}）\n True FDR: {FDR:.3f} (Target: {alpha})")
print(f"统计功效Power：{Power:.3f}\n Statistical Power: {Power:.3f}")
```
